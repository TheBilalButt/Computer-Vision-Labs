```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(f"OpenCV Version: {cv2.__version__}")
```

## 1. Feature Detection and Description with SIFT

```python
# Synthetic Base Canvas and Warped Pair
base = np.zeros((300, 300), dtype=np.uint8)
cv2.rectangle(base, (50, 50), (200, 200), 255, -1)
cv2.circle(base, (120, 120), 40, 0, -1)

# Affine warp
M = cv2.getRotationMatrix2D((150, 150), 30, 0.9)
target = cv2.warpAffine(base, M, (300, 300))

sift = cv2.SIFT_create()
kp1, des1 = sift.detectAndCompute(base, None)
kp2, des2 = sift.detectAndCompute(target, None)

print(f"Base Keypoints: {len(kp1)} | Target Keypoints: {len(kp2)}")
```

## 2. FLANN Matching and Ratio Test

```python
index_params = dict(algorithm=1, trees=5)
search_params = dict(checks=50)
flann = cv2.FlannBasedMatcher(index_params, search_params)

matches = flann.knnMatch(des1, des2, k=2)
good_matches = [m for m, n in matches if m.distance < 0.75 * n.distance]
print(f"Good Matches Retained: {len(good_matches)}")
```
