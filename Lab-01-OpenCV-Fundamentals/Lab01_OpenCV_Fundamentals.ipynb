```python
# 1. Environment Setup & Imports
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(f"OpenCV Version: {cv2.__version__}")
print(f"NumPy Version: {np.__version__}")
```

## 2. Generating Synthetic Test Image
Creating a multi-colored synthetic test pattern for color space transformations.

```python
# Create a 300x300 RGB test canvas
img = np.zeros((300, 300, 3), dtype=np.uint8)
img[0:150, 0:150] = [229, 9, 20]     # Red
img[0:150, 150:300] = [30, 144, 255] # Blue
img[150:300, 0:150] = [46, 204, 113] # Green
img[150:300, 150:300] = [241, 196, 15] # Yellow

plt.figure(figsize=(4, 4))
plt.imshow(img)
plt.title('Synthetic Test Image (RGB)')
plt.axis('off')
plt.show()
```

## 3. Color Space Conversions
OpenCV stores images in BGR format by default. Converting between color spaces is fundamental for computer vision pipelines.

```python
# Convert RGB to Grayscale, HSV, and LAB
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
axes[0].imshow(img); axes[0].set_title('Original RGB'); axes[0].axis('off')
axes[1].imshow(gray, cmap='gray'); axes[1].set_title('Grayscale'); axes[1].axis('off')
axes[2].imshow(hsv); axes[2].set_title('HSV Space'); axes[2].axis('off')
axes[3].imshow(lab); axes[3].set_title('LAB Space'); axes[3].axis('off')
plt.tight_layout()
plt.show()
```

## 4. Geometric Transformations
Resizing, translation, and rotational operations.

```python
# Rotation by 45 degrees
(h, w) = img.shape[:2]
center = (w // 2, h // 2)
M = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))

# Scaling
resized = cv2.resize(img, (150, 150), interpolation=cv2.INTER_AREA)

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(rotated); axes[0].set_title('Rotated 45°'); axes[0].axis('off')
axes[1].imshow(resized); axes[1].set_title('Downscaled 150x150'); axes[1].axis('off')
plt.show()
```

## 5. Histograms and Thresholding

```python
# Otsu Thresholding
_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

plt.figure(figsize=(6, 3))
plt.subplot(1, 2, 1); plt.imshow(gray, cmap='gray'); plt.title('Grayscale'); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(thresh, cmap='gray'); plt.title('Otsu Binary Threshold'); plt.axis('off')
plt.tight_layout()
plt.show()
```
